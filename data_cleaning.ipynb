```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display all columns
pd.set_option('display.max_columns', None)


```

```python
df = pd.read_csv('Retail_Transactions_Dataset.csv')
# print (df) 

print (df.head(5))
print (df.info())
print (df.describe())
print (df.shape)
```

```python
df['Date'].dtype
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
print(df.info())
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = df['Date'].dt.quarter
df['Day_Name'] = df['Date'].dt.day_name()
```

```python
print(df.head())
```

```python
df['Avg_Basket_Value'] = (
    df['Total_Cost'] /
    df['Total_Items']
)
```

```python
df.to_csv(
    "retail_transactions_clean.csv",
    index=False
)
```

```python
print(df.shape)
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.head())
```

```python
df["Promotion"] = df["Promotion"].fillna("No Promotion")
```

```python
df["Promotional_Flag"] = np.where(
    df['Promotion'] == "No Promotion",
    'No',
    'Yes'
)
```

```python
df["Revenue_Band"] = pd.cut(
    df["Total_Cost"],
    bins=[0,50,100,150,200,float('inf')],
    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
df["Order_Size"] = pd.cut(
    df["Total_Items"],
    bins=[0,2,5,10,float('inf')],
    labels=[
        "Small",
        "Medium",
        "Large",
        "Bulk"
    ]
)
```

```python
df["Weekend"] = np.where(
    df["Day_Name"].isin(["Saturday","Sunday"]),
    "Weekend",
    "Weekday"
)
```

```python
df.to_csv(
    "retail_transactions_clean.csv",
    index=False
)
```

```python
# Revenue Trend Analysis

monthly_revenue = (
    df.groupby("Month")["Total_Cost"]
      .sum()
      .sort_values(ascending=False)
)

monthly_revenue
```

```python
season_revenue = (
    df.groupby("Season")["Total_Cost"]
      .sum()
      .sort_values(ascending=False)
)

season_revenue
```

```python
city_revenue = (
    df.groupby("City")["Total_Cost"]
      .sum()
      .sort_values(ascending=False)
)

city_revenue.head(10)
```

```python
store_revenue = (
    df.groupby("Store_Type")["Total_Cost"]
      .sum()
      .sort_values(ascending=False)
)

store_revenue
```

```python
customer_segment = (
    df.groupby("Customer_Category")["Total_Cost"]
      .sum()
      .sort_values(ascending=False)
)

customer_segment
```

```python
promotion_revenue = (
    df.groupby("Promotion")["Total_Cost"]
      .agg(
          Revenue="sum",
          Avg_Order_Value="mean",
          Transactions="count"
      )
      .sort_values(
          by="Revenue",
          ascending=False
      )
)

promotion_revenue
```

```python
product_sales = (
    df.groupby("Product")["Total_Cost"]
      .sum()
      .sort_values(ascending=False)
)

product_sales.head(20)
```
